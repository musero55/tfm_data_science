```python
from pathlib import Path
import pandas as pd
import numpy as np

# 1. Ruta para guardar el archivo procesado
ruta_datos_simulados = Path('../../notebooks/fase_2_feature_engineering')
ruta_datos_procesados = Path('../../data/processed')

# Lista oficial de tus 33 países
paises = [
    'AUS', 'AUT', 'BEL', 'CAN', 'CHE', 'CHL', 'CZE', 'DEU', 'DNK', 'ESP', 
    'EST', 'FIN', 'FRA', 'GBR', 'GRC', 'HUN', 'IRL', 'ISL', 'ISR', 'ITA', 
    'JPN', 'KOR', 'MEX', 'NLD', 'NOR', 'NZL', 'POL', 'PRT', 'SVK', 'SVN', 
    'SWE', 'TUR', 'USA'
]
anios = [2012, 2015, 2018, 2022]
filas_aleatorias = []

# Establecemos semilla para que la aleatoriedad sea reproducible en tus pruebas
np.random.seed(42)

# 2. Generación de datos 100% aleatoria acotada
for pais in paises:
    for anio in anios:
        filas_aleatorias.append({
            'CNT': pais,
            'year': anio,
            # PIB per cápita aleatorio entre 15,000 y 90,000 USD
            'PIB_per_capita_PPP': round(np.random.uniform(15000, 90000), 2),
            # Gasto acumulado aleatorio entre 20,000 y 150,000 USD
            'Gasto_Alumno_PPP': round(np.random.uniform(20000, 150000), 2),
            # % de inversión en educación aleatorio entre 3.0% y 7.5%
            'Gasto_Educacion_PIB': round(np.random.uniform(3.0, 7.5), 2),
            # Índice Gini aleatorio entre 23.0 y 48.0 puntos
            'Indice_Gini': round(np.random.uniform(23.0, 48.0), 2),
            # Índice ESCS agrupado aleatorio entre -1.5 y 1.5 (escala estándar PISA)
            'ESCS_media_pais_anual': round(np.random.uniform(-1.5, 1.5), 2)
        })

df_macro_aleatorio = pd.DataFrame(filas_aleatorias)

# 3. Guardamos la matriz base en formato Parquet
df_macro_aleatorio.to_parquet(ruta_datos_simulados/ 'pisa_datos_economicos_simulados.parquet', index=False)
print(f"Matriz base aleatoria generada con éxito ({df_macro_aleatorio.shape[0]} registros).")
```
