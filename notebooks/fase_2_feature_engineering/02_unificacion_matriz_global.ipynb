```python
import pandas as pd
import os
```

```python
ruta_datos_intermedios ='../../data/intermediate'
ruta_datos_procesados ='../../data/processed'
```

Se debe realizar la lectura de los dataFrames que contienen los datos de los resultados de matemáticas medis por pais de los informes PISA y los datos de los indices derivados medios de cada pais

```python
ruta_datos_matematicas = os.path.join(
    ruta_datos_intermedios,
    'pisa22_math_pais_consolidado.parquet'
    )

ruta_datos_indices_pais = os.path.join(
    ruta_datos_intermedios,
    'pisa22_submuestra_paises.parquet'
    )
```

```python
#Lectura de las medias de matemáticas por paises
df_resultados_matematicas = pd.read_parquet(ruta_datos_matematicas)

#Lectura de las medias de los indices derivados por paises
df_indices_paises = pd.read_parquet(ruta_datos_indices_pais).reset_index()
```

```python
df_resultados_matematicas.head()
```

```python
df_indices_paises.head()
```

Se descarla la varianza de imputacion de las medias de matematicas al no se de utilizad en el estudio final

```python
df_resultados_matematicas.drop(columns='varianza_imputacion',inplace=True)
```

Realizamos la union de amnos dataframes por el código del pais

```python
df_analisis_global = pd.merge(
    df_resultados_matematicas,
    df_indices_paises,
    on='CNT',
    how='right'    
)
```

```python
df_analisis_global.head()
```

```python
parquet_submuestra_pais_consolidada=os.path.join(ruta_datos_procesados,'pisa22_submuestra_pais_consolidada.parquet')
```

```python
df_analisis_global.to_parquet(parquet_submuestra_pais_consolidada)
```
