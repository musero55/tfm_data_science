# Medias y deberes desde las tablas maestras

Se calculan aquí, desde `data/raw/datos_anuales` (muestra completa, todas las columnas), las medias de las asignaturas y los indicadores de deberes por estrato y por país. Sustituye a la lectura de `data/processed` (`pisa_medias_asignaturas_*.parquet` y `deberes_*.parquet`), que salía de ficheros de `data/raw` que ya no existen.

La lógica es la de los notebooks `01_procesamiento_asignaturas_PISA` y `01_procesamiento_deberes_PISA`, con una diferencia: **en 2022 `dispone_personal_apoyo` se recodifica (1 = sí → 1, 2 = no → 0) como en 2015 y 2018**; antes se promediaba el código original (1-2). El codebook oficial de PISA lo confirma para las tres ediciones.

```python
import pyarrow.parquet as pq

ruta_datos_intermedios = Path('../../data/intermediate')
ruta_datos_anuales = Path('../../data/raw') / 'datos_anuales'   # tablas maestras (en raw, no en intermediate)
ediciones_pisa = [2012, 2015, 2018, 2022]
claves_estrato = ['CNT', 'STRATUM']


def leer_maestra(año, tabla, columnas):
    """Columnas de la tabla maestra de `estudiantes` o `colegios` de una edición."""
    return pd.read_parquet(ruta_datos_anuales / f'pisa{año}_{tabla}.parquet', columns=columnas)
```

## Notas medias por estrato y por país

```python
# Media ponderada (W_FSTUWT) de los valores plausibles de cada asignatura, por estrato y por país
medias_estrato, medias_pais = [], []
for año in ediciones_pisa:
    nombres = pq.ParquetFile(ruta_datos_anuales / f'pisa{año}_estudiantes.parquet').schema.names
    columnas_pv = {dominio: [columna for columna in nombres if columna.startswith('PV') and columna.endswith(dominio)] for dominio in ('MATH', 'READ', 'SCIE')}
    estudiantes = leer_maestra(año, 'estudiantes', claves_estrato + ['W_FSTUWT'] + sum(columnas_pv.values(), []))

    por_estrato = [puntuaciones_pisa_estratos(estudiantes, dominio, 'W_FSTUWT') for dominio in columnas_pv]
    por_pais = [puntuaciones_pisa_pais(estudiantes, dominio, 'W_FSTUWT') for dominio in columnas_pv]
    medias_estrato.append(por_estrato[0].merge(por_estrato[1], on=claves_estrato).merge(por_estrato[2], on=claves_estrato).assign(year=año))
    medias_pais.append(por_pais[0].merge(por_pais[1], on='CNT').merge(por_pais[2], on='CNT').assign(year=año))

df_medias_estrato = pd.concat(medias_estrato, ignore_index=True)
df_medias_pais = pd.concat(medias_pais, ignore_index=True)
print(f'medias por estrato: {df_medias_estrato.shape} | por país: {df_medias_pais.shape}')
```

## Indicadores de deberes

Para cada edición se construyen `tiempo_deberes_totales_diarios` (alumno) y `dispone_personal_apoyo` (centro). Cada edición pregunta distinto, así que cada una tiene su celda; después se agregan igual para todas.

```python
origen_deberes = {}   # edición -> (estudiantes, colegios) con las dos variables ya calculadas

# --- 2012: horas semanales de tres ítems entre 5 días; apoyo = proxy con clases extra de matemáticas
estudiantes_deberes = leer_maestra(2012, 'estudiantes', claves_estrato + ['CNTSCHID', 'W_FSTUWT', 'ST57Q01', 'ST57Q02', 'ST57Q05'])
estudiantes_deberes['tiempo_deberes_totales_diarios'] = estudiantes_deberes[['ST57Q01', 'ST57Q02', 'ST57Q05']].sum(axis=1, min_count=1) / 5
colegios_deberes = leer_maestra(2012, 'colegios', claves_estrato + ['CNTSCHID', 'SC20Q01', 'SC21Q05'])
colegios_deberes['dispone_personal_apoyo'] = np.where((colegios_deberes['SC20Q01'] == 1) & colegios_deberes['SC21Q05'].isin([2, 3]), 1, 0)
colegios_deberes.loc[colegios_deberes['SC20Q01'].isna(), 'dispone_personal_apoyo'] = np.nan   # sin respuesta: se mantiene el NaN
origen_deberes[2012] = (estudiantes_deberes, colegios_deberes)
```

```python
# --- 2015: horas semanales de tres ítems (ST071Q01-03NA) entre 5; apoyo = SC052Q02NA (1 = sí, 2 = no)
columnas_tiempo_2015 = [f'ST071Q0{numero_item}NA' for numero_item in range(1, 4)]
estudiantes_deberes = leer_maestra(2015, 'estudiantes', claves_estrato + ['CNTSCHID', 'W_FSTUWT'] + columnas_tiempo_2015)
estudiantes_deberes['tiempo_deberes_totales_diarios'] = estudiantes_deberes[columnas_tiempo_2015].sum(axis=1, min_count=1) / 5
colegios_deberes = leer_maestra(2015, 'colegios', claves_estrato + ['CNTSCHID', 'SC052Q02NA'])
colegios_deberes['dispone_personal_apoyo'] = colegios_deberes['SC052Q02NA'].map({1: 1, 2: 0})
origen_deberes[2015] = (estudiantes_deberes, colegios_deberes)
```

```python
# --- 2018: horas diarias antes y después del colegio (EC158Q01HA y EC159Q01HA), entre 0,25 y 4,5; apoyo = SC052Q02NA
estudiantes_deberes = leer_maestra(2018, 'estudiantes', claves_estrato + ['CNTSCHID', 'W_FSTUWT', 'EC158Q01HA', 'EC159Q01HA'])
horas_manana = (estudiantes_deberes['EC158Q01HA'].where(estudiantes_deberes['EC158Q01HA'] < 10, np.nan) - 1).clip(lower=0)   # los códigos >= 10 no son válidos
horas_tarde = (estudiantes_deberes['EC159Q01HA'].where(estudiantes_deberes['EC159Q01HA'] < 10, np.nan) - 1).clip(lower=0)
estudiantes_deberes['tiempo_deberes_totales_diarios'] = (horas_manana + horas_tarde).clip(lower=0.25, upper=4.5)
colegios_deberes = leer_maestra(2018, 'colegios', claves_estrato + ['CNTSCHID', 'SC052Q02NA'])
colegios_deberes['dispone_personal_apoyo'] = colegios_deberes['SC052Q02NA'].map({1: 1, 2: 0})
origen_deberes[2018] = (estudiantes_deberes, colegios_deberes)
```

```python
# --- 2022: tramos de tiempo diario (ST296Q04JA) pasados a horas; apoyo = SC212Q02JA
mapeo_horas = {1: 0.25, 2: 0.75, 3: 1.50, 4: 2.50, 5: 3.50, 6: 4.50}   # <30 min, 30 min-1 h, 1-2 h, 2-3 h, 3-4 h, >4 h
estudiantes_deberes = leer_maestra(2022, 'estudiantes', claves_estrato + ['CNTSCHID', 'W_FSTUWT', 'ST296Q04JA'])
estudiantes_deberes['tiempo_deberes_totales_diarios'] = estudiantes_deberes['ST296Q04JA'].map(mapeo_horas)
colegios_deberes = leer_maestra(2022, 'colegios', claves_estrato + ['CNTSCHID', 'SC212Q02JA'])
# SC212Q02JA: 1 = Yes, 2 = No (codebook PISA 2022). Se recodifica como en 2015 y 2018 para que la media sea la proporción de centros con apoyo (0-1)
colegios_deberes['dispone_personal_apoyo'] = colegios_deberes['SC212Q02JA'].map({1: 1, 2: 0})
origen_deberes[2022] = (estudiantes_deberes, colegios_deberes)
```

```python
# Agregación igual para todas las ediciones: media ponderada por estrato y por país.
# Alumnos: peso W_FSTUWT. Centros: peso = suma de W_FSTUWT de sus alumnos.
deberes_estrato, deberes_pais = [], []
for año, (estudiantes_deberes, colegios_deberes) in origen_deberes.items():
    claves_centro = claves_estrato + ['CNTSCHID']
    pesos = estudiantes_deberes.groupby(claves_centro, as_index=False)['W_FSTUWT'].sum()
    colegios_deberes = colegios_deberes.merge(pesos, on=claves_centro, how='left', validate='one_to_one')
    for claves, salida in ((claves_estrato, deberes_estrato), (['CNT'], deberes_pais)):
        horas = promedio_ponderado_indicadores_agrupado(estudiantes_deberes[claves + ['tiempo_deberes_totales_diarios', 'W_FSTUWT']], ['tiempo_deberes_totales_diarios'], 'W_FSTUWT', claves)
        apoyo = promedio_ponderado_indicadores_agrupado(colegios_deberes[claves + ['dispone_personal_apoyo', 'W_FSTUWT']], ['dispone_personal_apoyo'], 'W_FSTUWT', claves)
        salida.append(horas.merge(apoyo, on=claves, how='left', validate='one_to_one').assign(year=año))

df_deberes_strat = pd.concat(deberes_estrato, ignore_index=True)[claves_estrato + ['year', 'tiempo_deberes_totales_diarios', 'dispone_personal_apoyo']]
df_deberes_cnt = pd.concat(deberes_pais, ignore_index=True)[['CNT', 'year', 'tiempo_deberes_totales_diarios', 'dispone_personal_apoyo']]
print(f'deberes por estrato: {df_deberes_strat.shape} | por país: {df_deberes_cnt.shape}')
df_deberes_strat.groupby('year')[['tiempo_deberes_totales_diarios', 'dispone_personal_apoyo']].agg(['count', 'min', 'mean', 'max']).round(2)
```

```python
# Comprobación opcional frente a `data/processed/deberes_strat.parquet` (solo si existe): deben coincidir salvo `dispone_personal_apoyo` de 2022
ruta_antiguo = ruta_datos_procesados / 'deberes_strat.parquet'
if ruta_antiguo.exists():
    antiguo = pd.read_parquet(ruta_antiguo)
    comparacion = df_deberes_strat.merge(antiguo, on=claves_estrato + ['year'], suffixes=('', '_antiguo'), how='outer', indicator=True)
    print('estratos solo en uno de los dos:', int((comparacion['_merge'] != 'both').sum()))
    for variable in ('tiempo_deberes_totales_diarios', 'dispone_personal_apoyo'):
        diferencia_absoluta = (comparacion[variable] - comparacion[f'{variable}_antiguo']).abs()
        print(f'{variable}: diferencia máxima por edición', diferencia_absoluta.groupby(comparacion['year']).max().round(6).to_dict())
    media_guardada_2022 = antiguo.loc[antiguo['year'] == 2022, 'tiempo_deberes_totales_diarios'].mean()
    media_horas_2022 = df_deberes_strat.loc[df_deberes_strat['year'] == 2022, 'tiempo_deberes_totales_diarios'].mean()
    codigos_2022 = promedio_ponderado_indicadores_agrupado(origen_deberes[2022][0][claves_estrato + ['ST296Q04JA', 'W_FSTUWT']], ['ST296Q04JA'], 'W_FSTUWT', claves_estrato)
    comparacion_codigos = antiguo[antiguo['year'] == 2022].merge(codigos_2022, on=claves_estrato)
    diferencia_codigos = (comparacion_codigos['tiempo_deberes_totales_diarios'] - comparacion_codigos['ST296Q04JA']).abs().max()
    print(f'2022, media por estrato de las horas de deberes: guardada en data/processed = {media_guardada_2022:.2f}; calculada aquí = {media_horas_2022:.2f}; '
          f'diferencia máxima entre lo guardado y la media de los códigos de categoría 1-6 = {diferencia_codigos:.1e}')
else:
    print('No existe data/processed/deberes_strat.parquet: no se compara.')
```

**Diferencias esperadas** (solo en 2022; 2012, 2015 y 2018 coinciden):

* `dispone_personal_apoyo`: aquí está recodificada (0-1); en `data/processed/deberes_strat.parquet` era la media del código original (1-2).
* `tiempo_deberes_totales_diarios`: `data/processed/deberes_strat.parquet` trae la **media de los códigos de categoría (1-6)**, de media 2,66, y no las horas: el mapeo a horas de `01_procesamiento_deberes_PISA` no se aplicó en lo guardado (el valor coincide con la media de códigos con una diferencia máxima de 3,5·10⁻⁷, como imprime la celda anterior). Aquí se aplica ese mapeo (<30 min → 0,25 h … >4 h → 4,5 h) y la media de 2022 es 1,43 h.

## Feature engineering del presupuesto

A partir de las variables básicas (`presupuesto`) se calculan, una por una, las variables derivadas. Todos los importes están en USD PPP constantes de 2020, así que se pueden dividir entre sí.

```python
df_presupuesto = presupuesto.copy()

# 1. Cómo se gasta: composición del gasto de secundaria (% del gasto total = corriente + capital)
gasto_total = df_presupuesto['gasto_total_corriente_secundaria_usd_ppp'] + df_presupuesto['gasto_total_capital_secundaria_usd_ppp']
df_presupuesto['pct_gasto_secundaria_en_personal'] = 100 * df_presupuesto['gasto_total_personal_secundaria_usd_ppp'] / gasto_total
df_presupuesto['pct_gasto_secundaria_en_docentes'] = 100 * df_presupuesto['gasto_total_docentes_secundaria_usd_ppp'] / gasto_total
df_presupuesto['pct_gasto_secundaria_en_capital'] = 100 * df_presupuesto['gasto_total_capital_secundaria_usd_ppp'] / gasto_total

# Servicios auxiliares (transporte, comidas, alojamiento) sobre el gasto por alumno
df_presupuesto['servicios_auxiliares_pct_gasto_alumno_secundaria'] = 100 * df_presupuesto['gasto_alumno_secundaria_servicios_auxiliares_usd_ppp'] / df_presupuesto['gasto_alumno_secundaria_usd_ppp']

df_presupuesto[['pct_gasto_secundaria_en_personal', 'pct_gasto_secundaria_en_docentes', 'pct_gasto_secundaria_en_capital', 'servicios_auxiliares_pct_gasto_alumno_secundaria']].describe().T.round(1)
```

```python
# 2. Relativos a la renta del país (% del PIB per cápita)
df_presupuesto['gasto_alumno_secundaria_pct_pib_per_capita'] = 100 * df_presupuesto['gasto_alumno_secundaria_usd_ppp'] / df_presupuesto['pib_per_capita_usd_ppp']
df_presupuesto['salario_legal_docente_15_años_experiencia_pct_pib_per_capita'] = 100 * df_presupuesto['salario_legal_docente_15_años_experiencia_usd_ppp'] / df_presupuesto['pib_per_capita_usd_ppp']

df_presupuesto[['gasto_alumno_secundaria_pct_pib_per_capita', 'salario_legal_docente_15_años_experiencia_pct_pib_per_capita']].describe().T.round(1)
```

```python
# 3. Dotación y coste docente
df_presupuesto['docentes_por_100_alumnos'] = 100 / df_presupuesto['alumnos_por_docente_secundaria_primera_etapa']
# Salario de un docente repartido entre sus alumnos (USD PPP por alumno) y su peso en el gasto por alumno
df_presupuesto['coste_docente_por_alumno_usd_ppp'] = df_presupuesto['salario_legal_docente_15_años_experiencia_usd_ppp'] / df_presupuesto['alumnos_por_docente_secundaria_primera_etapa']
df_presupuesto['coste_docente_por_alumno_pct_gasto_alumno_secundaria'] = 100 * df_presupuesto['coste_docente_por_alumno_usd_ppp'] / df_presupuesto['gasto_alumno_secundaria_usd_ppp']

df_presupuesto[['docentes_por_100_alumnos', 'coste_docente_por_alumno_usd_ppp', 'coste_docente_por_alumno_pct_gasto_alumno_secundaria']].describe().T.round(1)
```

```python
# 4. Reparto entre niveles (gasto por alumno relativo a secundaria)
df_presupuesto['ratio_gasto_preprimaria_sobre_secundaria'] = df_presupuesto['gasto_alumno_preprimaria_usd_ppp'] / df_presupuesto['gasto_alumno_secundaria_usd_ppp']

df_presupuesto[['ratio_gasto_preprimaria_sobre_secundaria']].describe().T.round(2)
```

```python
# 5. Años previos, con la serie anual (solo se interpolan linealmente los huecos interiores)
ancha = {columna_gasto: serie_anual.pivot(index='año', columns='pais', values=columna_gasto).interpolate(limit_area='inside')
         for columna_gasto in ('gasto_alumno_primaria_usd_ppp', 'gasto_alumno_secundaria_primera_etapa_usd_ppp', 'gasto_alumno_secundaria_usd_ppp')}
media_4_años_por_edicion, acumulado = {}, {}
for edicion in sorted(df_presupuesto['edicion'].unique()):
    # Media del gasto por alumno de los 4 años hasta la edición (t-3 a t), con al menos 3 años de dato
    ventana = ancha['gasto_alumno_secundaria_usd_ppp'].loc[edicion - 3:edicion]
    media_4_años_por_edicion[edicion] = ventana.mean().where(ventana.notna().sum() >= 3)
    # Gasto acumulado de los 6 a los 15 años: 6 años de primaria (t-9 a t-4) + 4 de secundaria 1.ª etapa (t-3 a t)
    primaria, secundaria = ancha['gasto_alumno_primaria_usd_ppp'].loc[edicion - 9:edicion - 4], ancha['gasto_alumno_secundaria_primera_etapa_usd_ppp'].loc[edicion - 3:edicion]
    completo = (primaria.notna().sum() == 6) & (secundaria.notna().sum() == 4)
    acumulado[edicion] = (primaria.sum() + secundaria.sum()).where(completo)

df_presupuesto['gasto_alumno_secundaria_media_4_años_usd_ppp'] = [media_4_años_por_edicion[edicion_fila].get(pais_fila, np.nan) for pais_fila, edicion_fila in zip(df_presupuesto['pais'], df_presupuesto['edicion'])]
df_presupuesto['gasto_acumulado_alumno_6_a_15_años_usd_ppp'] = [acumulado[edicion_fila].get(pais_fila, np.nan) for pais_fila, edicion_fila in zip(df_presupuesto['pais'], df_presupuesto['edicion'])]

df_presupuesto[['gasto_alumno_secundaria_media_4_años_usd_ppp', 'gasto_acumulado_alumno_6_a_15_años_usd_ppp']].describe().T.round(0)
```

```python
# ---------------------------------------------------------------------
# 2. ESCS REAL POR PAÍS Y POR ESTRATO (tablas maestras)
# ---------------------------------------------------------------------
def escs_por_pais_y_estrato(año):
    """ESCS_TREND medio ponderado con W_FSTUWT, por estrato y por país."""
    estudiantes = pd.read_parquet(ruta_datos_anuales / f'pisa{año}_estudiantes.parquet', columns=['CNT', 'STRATUM', 'W_FSTUWT', 'ESCS_TREND']).dropna()
    estudiantes['STRATUM'] = estudiantes['STRATUM'].astype(str)
    estudiantes['_wx'] = estudiantes['ESCS_TREND'] * estudiantes['W_FSTUWT']
    estrato, pais = (estudiantes.groupby(claves_agrupacion)[['_wx', 'W_FSTUWT']].sum() for claves_agrupacion in (['CNT', 'STRATUM'], ['CNT']))
    return ((estrato['_wx'] / estrato['W_FSTUWT']).rename('escs_estrato').reset_index().assign(year=año),
            (pais['_wx'] / pais['W_FSTUWT']).rename('escs_media_pais').reset_index().assign(year=año))


resultados = [escs_por_pais_y_estrato(edicion_pisa) for edicion_pisa in (2012, 2015, 2018, 2022)]
df_escs_estrato = pd.concat([resultado[0] for resultado in resultados], ignore_index=True)
df_escs_pais = pd.concat([resultado[1] for resultado in resultados], ignore_index=True)
print(f'ESCS por estrato: {df_escs_estrato.shape} | por país: {df_escs_pais.shape}')

# ---------------------------------------------------------------------
# 3. UNIÓN A: TABLA LARGA POR ESTRATOS
# ---------------------------------------------------------------------
for panel in (df_panel_estrato, df_panel_pais):
    panel['year'] = panel['year'].astype(int)
df_panel_estrato['STRATUM'] = df_panel_estrato['STRATUM'].astype(str)

macro_pais = df_presupuesto[['CNT', 'year'] + variables_pais].merge(df_escs_pais, on=['CNT', 'year'], how='left')
columnas_macro = variables_pais + ['escs_media_pais']

df_panel_estrato_final = (df_panel_estrato
                          .merge(macro_pais, on=['CNT', 'year'], how='left')
                          .merge(df_escs_estrato, on=['CNT', 'STRATUM', 'year'], how='left'))
columnas_macro_estrato = columnas_macro + ['escs_estrato']
df_panel_estrato_final = df_panel_estrato_final[[columna for columna in df_panel_estrato_final.columns if columna not in columnas_macro_estrato] + columnas_macro_estrato]

print("=== 1. PANEL POR ESTRATO FINAL COMPLETO (TABLA LARGA) ===")
display(df_panel_estrato_final.head())
print(f'Dimensiones del panel largo: {df_panel_estrato_final.shape}')
print('estratos sin ESCS:', int(df_panel_estrato_final['escs_estrato'].isna().sum()))
```

```python
# ---------------------------------------------------------------------
# UNIÓN B: TABLA POR PAÍS
# ---------------------------------------------------------------------
df_pais_transaccional = df_panel_pais.merge(macro_pais, on=['CNT', 'year'], how='left')

# Pivotamos: países en filas y años en columnas
columnas_a_pivotar = [col for col in df_pais_transaccional.columns if col not in ['CNT', 'year']]
df_pais_ancho = df_pais_transaccional.pivot(index='CNT', columns='year', values=columnas_a_pivotar)
df_pais_ancho.columns = [f'{var}_{anio}' for var, anio in df_pais_ancho.columns]
df_pais_ancho = df_pais_ancho.reset_index()

print("=== 2. PANEL POR PAÍS FINAL COMPLETO (FORMATO ANCHO PIVOTADO) ===")
display(df_pais_ancho.head())
print(f'Dimensiones del panel ancho: {df_pais_ancho.shape}')
```

```python
def ordenar_columnas_panel(df):
    claves = [col for col in ['CNT', 'STRATUM', 'year'] if col in df.columns]
    medias = [
        col for col in [
            'media_math_pais', 'media_read_pais', 'media_scie_pais',
            'media_math_estrato', 'media_read_estrato', 'media_scie_estrato',
        ]
        if col in df.columns
    ]
    resto = [col for col in df.columns if col not in claves + medias]
    return df[claves + medias + resto]


# Paneles en formato largo
df_panel_estrato = ordenar_columnas_panel(df_panel_estrato)
df_panel_estrato_final = ordenar_columnas_panel(df_panel_estrato_final)
df_panel_pais = ordenar_columnas_panel(df_panel_pais)
df_pais_transaccional = ordenar_columnas_panel(df_pais_transaccional)


# Panel por país en formato ancho: medias primero, agrupadas por asignatura
if 'df_pais_ancho' in globals():
    orden_medias = [
        col
        for prefijo in ['media_math_pais_', 'media_read_pais_', 'media_scie_pais_']
        for col in df_pais_ancho.columns
        if col.startswith(prefijo)
    ]
    claves = ['CNT'] if 'CNT' in df_pais_ancho.columns else []
    resto = [
        col for col in df_pais_ancho.columns
        if col not in claves + orden_medias
    ]
    df_pais_ancho = df_pais_ancho[claves + orden_medias + resto]
```

## Cuánta variación hay en cada variable

Cada variable del panel por estrato varía a un nivel distinto: las de presupuesto, PIB y Gini son del **país y la edición** (todos sus estratos comparten el valor), mientras que las notas, el `ESCS`, los deberes y el personal de apoyo se calculan **por estrato**. Para cada una se reparte la varianza total (cada fila del panel pesa igual) en tres partes que suman 100 %:

* **entre países**: diferencias entre las medias de los países;
* **entre ediciones**: cuánto se separa la media de un país en una edición de su media en las cuatro;
* **entre estratos**: lo que varía entre los estratos de un mismo país y edición.

Los estratos de 2022 tienen otro código que los de 2012-2018, por lo que parte de la variación «entre estratos» incluye cambios de definición de los estratos.

```python
def descomposicion_varianza(panel, variables):
    """% de la varianza de cada variable entre países, entre ediciones (dentro del país) y entre estratos (dentro de país y edición)."""
    filas = {}
    for variable in variables:
        datos_variable = panel[['CNT', 'year', variable]].dropna()
        total = ((datos_variable[variable] - datos_variable[variable].mean()) ** 2).sum()
        media_pais = datos_variable.groupby('CNT')[variable].transform('mean')
        media_pais_edicion = datos_variable.groupby(['CNT', 'year'])[variable].transform('mean')
        filas[variable] = {'entre países': ((media_pais - datos_variable[variable].mean()) ** 2).sum() / total,
                    'entre ediciones': ((media_pais_edicion - media_pais) ** 2).sum() / total,
                    'entre estratos': ((datos_variable[variable] - media_pais_edicion) ** 2).sum() / total}
    return 100 * pd.DataFrame(filas).T


variables_estrato = ['media_math_estrato', 'escs_estrato', 'tiempo_deberes_totales_diarios', 'dispone_personal_apoyo']
variables_pais_edicion = ['escs_media_pais', 'pib_per_capita_usd_ppp', 'gasto_alumno_secundaria_usd_ppp', 'gasto_pct_pib_primaria_a_postsecundaria', 'financiacion_privada_pct_gasto_secundaria', 'gini']
reparto_varianza = descomposicion_varianza(df_panel_estrato_final, variables_estrato + variables_pais_edicion)

fig, ax = plt.subplots(figsize=(10, 5))
acumulado = pd.Series(0.0, index=reparto_varianza.index)
for componente, color in zip(reparto_varianza.columns, ['tab:blue', 'tab:orange', 'tab:green']):
    barras = ax.barh(reparto_varianza.index, reparto_varianza[componente], left=acumulado, color=color, label=componente)
    ax.bar_label(barras, labels=[f'{porcentaje:.0f} %' if porcentaje >= 4 else '' for porcentaje in reparto_varianza[componente]], label_type='center', fontsize=8, color='white')
    acumulado += reparto_varianza[componente]
ax.axhline(len(variables_estrato) - 0.5, color='black', lw=0.8, ls='--')
ax.invert_yaxis()
ax.set_xlabel('% de la varianza')
ax.set_title('Dónde está la variación de cada variable (arriba: por estrato; debajo de la línea: del país y la edición)', fontsize=10)
ax.legend(ncol=3, loc='lower center', bbox_to_anchor=(0.5, -0.25), fontsize=8)
plt.tight_layout()
plt.show()
```

### Qué implica para el modelo

* **Presupuesto, PIB y Gini son contexto de país.** Más del 90 % de su variación es entre países (gasto por alumno 91 %, financiación privada 93 %, Gini 98 %) y ninguna entre estratos: los estratos de un país y edición repiten el mismo valor. Su información equivale a unos 33 países, no a 2 328 filas (muestra efectiva: `01_EDA_presupuestos.ipynb`), así que repetirlos en cada estrato no aumenta la muestra y con ellos solo se pueden estudiar asociaciones entre países, no el efecto de cambiar el presupuesto.
* **Las variables de estrato sí varían dentro del país** y esa es la variación con la que el modelo puede discriminar entre filas de un mismo grupo de países: el 64 % de `dispone_personal_apoyo`, el 66 % de la nota media, el 38 % del `ESCS` y el 37 % de las horas de deberes es entre estratos. Por eso la tabla por estratos usa `escs_estrato` y no solo la media del país (`escs_media_pais`, 95 % entre países).
* **El `ESCS` casi no cambia entre ediciones** (6 %), a diferencia de los deberes (44 %). En los deberes ese cambio no es real: cada edición mide un constructo distinto (ver «Indicadores de deberes»).
* **Consecuencias**: el perfil macroeconómico y el presupuesto sitúan al estrato en un grupo de países; lo que distingue a los estratos dentro de ese grupo son el `ESCS`, los deberes y el apoyo. Al validar, conviene separar por **país** y no por filas, porque los estratos de un mismo país se parecen entre sí.

## Calculo de la presión académica


Para evaluar el impacto real de la carga de trabajo en casa, se crea una nueva métrica continua denominada **Presión Académica Neta**. Esta variable relaciona directamente la cantidad de deberes con los recursos de apoyo que ofrece el centro escolar, bajo la premisa de que el apoyo reduce la carga escolar

**Fórmula:**
$$Presion\_Academica\_Neta = \frac{tiempo\_deberes\_totales\_diarios}{1 + dispone\_personal\_apoyo}$$

** Denominador ($1 + apoyo$):**
 (Efecto amortiguador):** El denominador actúa como un divisor de la carga de trabajo. 
  * Si el colegio **no tiene apoyo** (0), el denominador es $1 + 0 = 1$. El alumno absorbe el 100% de la carga de los deberes.
  * Si el colegio **sí tiene apoyo** (1), el denominador es $1 + 1 = 2$. La presión neta se reduce a la mitad, reflejando cómo los profesores de refuerzo alivian el esfuerzo solitario en casa.

```python
df_panel_estrato_final['Presion_Academica_Neta'] = df_panel_estrato_final['tiempo_deberes_totales_diarios'] / (1 + df_panel_estrato_final['dispone_personal_apoyo'])
```

## Discretización del Nivel Socioeconómico (ECS)

Se transforma el índice continuo de estatus socioeconómico (`escs_estrato`, `ESCS_TREND` medio del estrato) en una variable categórica ordinal. De tal forma que se analice el impacto de las variables educativas a través de un indice cualitativo mas intuitivo

**Regla de codificación:**
Se utiliza una discretización basada en cuantiles para dividir los estratos en tres grupos. De esta forma se pueden aprender patrones de cada grupo social

* **Bajo:** menor índice socioeconómico.
* **Medio:** punto intermedio del cuantil.
* **Alto:** mayor índice socioeconómico.

El `ESCS_TREND` es comparable entre 2012 y 2022, así que los cuantiles se calculan con todas las ediciones juntas. Esta segmentación permitirá evaluar interacciones complejas, como por ejemplo, si la *Presión Académica Neta* tiene un efecto dispar en el rendimiento dependiendo de los recursos del hogar y del entorno del estrato.

```python
# Etiquetas para los 3 niveles
etiquetas_socioeconomicas = ['Bajo', 'Medio', 'Alto']

# Dividimos la variable continua en 3 partes iguales ( q=3 )
df_panel_estrato_final['ECS_categoria'] = pd.qcut(
    df_panel_estrato_final['escs_estrato'], 
    q=3, 
    labels=etiquetas_socioeconomicas
)

# Visualizamos cómo ha quedado el reparto de estratos por cada nivel
print("Porcentaje de distribución de estratos por nivel socioeconómico:")
print(df_panel_estrato_final['ECS_categoria'].value_counts(normalize=True) * 100)
```

## Perfil Macroeconómico

Para dotar al modelo de un contexto estructural simplificado, se segmentan los países en perfiles macroeconómicos utilizando un algoritmo no supervisado (K-Means).

**Metodología:**
1. Se aísla un *dataframe* transversal a nivel de país con los indicadores estructurales: `pib_per_capita_usd_ppp`, `gasto_alumno_secundaria_usd_ppp`, `gasto_pct_pib_primaria_a_postsecundaria`, `financiacion_privada_pct_gasto_secundaria`, `gini` y `escs_media_pais`.
    * Creamos la "foto fija" de la década para cada país
    * Calcular la media de la década elimina el "ruido" de posibles picos o caídas puntuales en un año concreto y captura el verdadero modelo socioeconómico del país

2. Se escalan las variables (StandardScaler) para evitar que magnitudes como el PIB dominen el cálculo de distancias sobre porcentajes como la financiación privada.
3. Se entrena un modelo K-Means (previa evaluación del número óptimo de clústeres mediante el método del codo o silueta).
4. Se asigna la etiqueta resultante (`Perfil_Pais`) a la tabla longitudinal larga, mapeando cada estrato a la realidad estructural de su nación.

```python
# Foto fija de cada país: media de las ediciones disponibles. Se promedia sobre las filas país-edición (`macro_pais`),
# no sobre los estratos, para que cada edición pese igual aunque tenga más estratos
variables_macro = ['pib_per_capita_usd_ppp', 'gasto_alumno_secundaria_usd_ppp', 'gasto_pct_pib_primaria_a_postsecundaria', 'financiacion_privada_pct_gasto_secundaria', 'gini', 'escs_media_pais']
df_macro_pais = macro_pais.groupby('CNT')[variables_macro].mean()
print(f'{df_macro_pais.dropna().shape[0]} países con las {len(variables_macro)} variables; sin alguna: {sorted(df_macro_pais[df_macro_pais.isna().any(axis=1)].index)}')
df_macro_pais = df_macro_pais.dropna()
```

Mediante el `StandardScaler` transformamos los datos de cada columna para que adopten una distribución normal estándar ( media en 0 y desviacion 1)

Se obliga así al modelo a darle la misma importancia a cada variable. Esto es, las variaciones de las variables se encuentran en el mismo rango dado que K-means es muy sensible a las magnitudes

```python
#Instanciamos el estandar scaler
scaler = StandardScaler()

#transformamos los datos
datos_macro_escalados = scaler.fit_transform(df_macro_pais)
```

### Analisis de datos para clustering

```python
#Calculo de los modelos
inercias = []
siluetas = []

rango_k = range(2, 9) # Evaluamos desde 2 hasta 8 clústeres

# Bucle para probar diferentes valores de agrupaciones
for k in rango_k:
    modelo_k = KMeans(n_clusters=k, random_state=42, n_init='auto')
    etiquetas = modelo_k.fit_predict(datos_macro_escalados)
    
    inercias.append(modelo_k.inertia_)
    siluetas.append(silhouette_score(datos_macro_escalados, etiquetas))
```

```python
# Resumen por k: inercia, silueta y tamaño del grupo más pequeño
resumen_k = pd.DataFrame({'k': list(rango_k), 'inercia': inercias, 'silueta': siluetas}).set_index('k')
resumen_k['grupo más pequeño'] = [np.bincount(KMeans(n_clusters=numero_grupos, random_state=42, n_init='auto').fit_predict(datos_macro_escalados)).min() for numero_grupos in rango_k]
display(resumen_k.round(3))

# Criterio 1 (codo): el k más alejado de la recta que une el primer y el último punto de la curva de inercia
candidatos_k, inercias_array = np.array(list(rango_k), dtype=float), np.array(inercias)
recta = inercias_array[0] + (inercias_array[-1] - inercias_array[0]) * (candidatos_k - candidatos_k[0]) / (candidatos_k[-1] - candidatos_k[0])
k_codo = int(candidatos_k[np.argmax(recta - inercias_array)])

# Criterio 2 (silueta): el k con mayor silueta entre los que no dejan grupos de menos de 3 países
min_paises = 3
k_silueta = int(resumen_k[resumen_k['grupo más pequeño'] >= min_paises]['silueta'].idxmax())

k_optimo = k_silueta
print(f'k según el codo = {k_codo} | k según la silueta = {k_silueta} -> k elegido = {k_optimo}')

# Composición de los grupos para los k más pequeños
for numero_grupos in range(2, 6):
    etiquetas_k = KMeans(n_clusters=numero_grupos, random_state=42, n_init='auto').fit_predict(datos_macro_escalados)
    print(f'k={numero_grupos}: ' + ' | '.join(f"{', '.join(df_macro_pais.index[etiquetas_k == grupo])}" for grupo in range(numero_grupos)))
```

```python


# Configuración de la figura con dos gráficos en paralelo
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico del Método del Codo
ax[0].plot(rango_k, inercias, marker='o', color='steelblue', linewidth=2)
ax[0].set_title('Método del Codo (Inercia)')
ax[0].set_xlabel('Número de Clústeres (k)')
ax[0].set_ylabel('Inercia (Menor es más compacto)')
ax[0].grid(True, linestyle='--', alpha=0.5)

# Gráfico del Método de la Silueta
ax[1].plot(rango_k, siluetas, marker='s', color='darkorange', linewidth=2)
ax[1].set_title('Método de la Silueta')
ax[1].set_xlabel('Número de Clústeres (k)')
ax[1].set_ylabel('Coeficiente de Silueta (Mayor es mejor)')
ax[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

**NÚMERO ÓPTIMO**: se decide con dos criterios calculados en la celda anterior, no a mano. Ambos dan **k = 3**.

* **Método del Codo**: se toma el k más alejado de la recta que une el primer y el último punto de la curva de inercia. La inercia baja de 128,7 (k=2) a 85,9 (k=3) y a 71,8 (k=4); k=3 es el punto de mayor distancia a la recta.

* **Método de la Silueta**: es máxima en k=3 (0,30) entre los k que no dejan grupos de menos de 3 países (k=7 y k=8 ya dejan un grupo de un solo país).

La silueta es baja en todos los casos (~0,3): los perfiles son una descripción útil, no grupos nítidamente separados. Si cambian los datos, `k_optimo` se recalcula (los nombres de los perfiles habría que revisarlos).

```python
#Entrenamos el modelo final en función del numero de clusters encontrados
kmeans = KMeans(n_clusters=k_optimo, random_state=42, n_init='auto')
etiquetas = kmeans.fit_predict(datos_macro_escalados)

# Numeramos los grupos por gasto por alumno creciente, para que la etiqueta no dependa del azar de K-Means
orden = df_macro_pais.groupby(etiquetas)['gasto_alumno_secundaria_usd_ppp'].mean().sort_values().index
rango_grupo = {grupo_original: rango for rango, grupo_original in enumerate(orden)}
df_macro_pais['Perfil_Pais'] = [rango_grupo[grupo_original] for grupo_original in etiquetas]

# Países con alguna variable sin dato (p. ej. Suiza no tiene financiación privada): grupo más cercano
# según las variables que sí tienen (distancia media al centro, en valores escalados)
macro_incompleto = macro_pais.groupby('CNT')[variables_macro].mean()
macro_incompleto = macro_incompleto[macro_incompleto.isna().any(axis=1) & macro_incompleto.notna().any(axis=1)]
valores_escalados = ((macro_incompleto - scaler.mean_) / scaler.scale_).to_numpy()
centros = kmeans.cluster_centers_[list(orden)]
distancia = np.nanmean((valores_escalados[:, None, :] - centros[None, :, :]) ** 2, axis=2)
perfil_paises = pd.concat([df_macro_pais['Perfil_Pais'], pd.Series(distancia.argmin(axis=1), index=macro_incompleto.index)]).rename('Perfil_Pais')
print('asignados por cercanía:', dict(perfil_paises[macro_incompleto.index]))
```

```python
# Comprobación: ¿cambian los grupos si no se usa el Gini? (mismo k, mismas filas)
variables_sin_gini = [variable for variable in variables_macro if variable != 'gini']
datos_sin_gini = StandardScaler().fit_transform(df_macro_pais[variables_sin_gini])
etiquetas_sin_gini = KMeans(n_clusters=k_optimo, random_state=42, n_init='auto').fit_predict(datos_sin_gini)
pd.crosstab(df_macro_pais['Perfil_Pais'], etiquetas_sin_gini, rownames=['perfil con Gini'], colnames=['grupo sin Gini'])
```
