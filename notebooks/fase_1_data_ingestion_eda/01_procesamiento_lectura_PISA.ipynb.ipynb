# Datos resultados lectura PISA

```python
columnas_lectura = [col for col in meta.column_names if 'PV' and 'READ' in col]
```

```python
print(f"Columnas encontradas: {len(columnas_lectura)}")
print(columnas_lectura)
```

```python
#Añadimos el resto de columnas de interes para realizar el estudio : 
columnas_interes = ['CNT', 'W_FSTUWT'] + columnas_lectura
```

```python
df_pisa, meta_real = pyreadstat.read_sas7bdat(
    '../../data/raw/CY08MSP_STU_QQQ.SAS7BDAT', 
    usecols=columnas_interes
)
```

```python
df_pisa.info(memory_usage='deep')
```

```python
display(df_pisa.head())
display(df_pisa.tail())
```

```python
rendimiento_lectura_PISA= consolidar_puntuaciones_pisa(df_pisa,competencia='READ')
```

```python
medias_pv_lectura=[]
for i in range(1, 11):
    col_actual = f'PV{i}READ'
    peso_final = 'W_FSTUWT'
    
    # Calcular la media ponderada agrupada por pais (CNT)
    media_ponderada_pv = df_pisa.groupby('CNT').apply(
        lambda x: (x[col_actual] * x[peso_final]).sum() / x[peso_final].sum(),
        include_groups=False
    )
    medias_pv_lectura.append(media_ponderada_pv)
```

```python
df_lectura_consolidado = pd.concat(medias_pv_lectura,axis=1)
```

```python
#Este es el dataframe que contiene cada media de los PV de lectura independiente
df_lectura_consolidado.head()
```

```python
#Hacemos la media de cada fila para obtener el resultado final
puntuacion_lectura = df_lectura_consolidado.mean(axis='columns') #Media
error_imputacion_pais = df_lectura_consolidado.var(axis=1)       #Varianza
```

```python
rendimiento_lectura_PISA =pd.DataFrame({
    'media_lectura_pisa': puntuacion_lectura,
    'varianza_imputacion': error_imputacion_pais
}).reset_index()
```

```python
rendimiento_lectura_PISA.shape
```

```python
lectura_top_cnt=rendimiento_lectura_PISA.sort_values(by='media_lectura_pisa',ascending=False)
```
