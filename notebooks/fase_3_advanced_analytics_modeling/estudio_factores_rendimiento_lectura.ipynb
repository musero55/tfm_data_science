# Obtención de datos

En la siguiente tabla se resumen las features que obtendremos para hacer el estudio de factores que pueden afectar a los resultados de lectura de un pais


| Archivo de Origen | Claves de Vinculación | Variables Clave a Extraer | Información que Aporta |
| :--- | :--- | :--- | :--- |
| **STU_QQQ**<br>*(Cuestionario del Estudiante)* | `CNT` (País)<br>`SCHOOLID` (ID Escuela)<br>`STUDENTID` (ID Alumno) | • `PV1READ` a `PV10READ` (Lectura)<br>• `W_FSTUWT` (Pesos finales)<br>• `ESCS` (Índice Socioeconómico)<br>• `DISCLIMA` (Clima de Disciplina)<br>• `BELONG` (Sentido de Pertenencia)<br>• `REPEAT` o `GRADE_REP` (Repetición) | Es el núcleo del rendimiento de los alumnos. Aporta las variables demográficas, socioeconómicas, de trayectoria escolar y de percepción actitudinal dentro del aula. |
| **SCH_QQQ**<br>*(Cuestionario de la Escuela)* | `CNT` (País)<br>`SCHOOLID` (ID Escuela) | • `SCHAUTO` (Autonomía escolar)<br>• `RESPCUR` (Autonomía curricular)<br>• `RESPRES` (Autonomía en recursos)<br>• `STAFFSHORT` (Escasez de personal)<br>• `MATEDU` (Escasez de recursos materiales) | Aporta la perspectiva institucional y de contexto escolar: gobernanza, capacidad de toma de decisiones, políticas internas del centro y limitaciones de infraestructura física o humana. |
| **CNT_QQQ / Macro**<br>*(Recursos y Criterios a nivel País)* | `CNT` (País) | • Gasto acumulado por alumno<br>• PIB per cápita del país<br>*(Indicadores macroeconómicos de la OCDE)* | Permite añadir variables contextuales a nivel de sistema educativo nacional. Sirve para controlar el impacto de la riqueza del país y la inversión financiera directa en educación. |
